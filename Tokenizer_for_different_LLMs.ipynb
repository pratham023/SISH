```python
#BPE tokenizer
import tiktoken

enc = tiktoken.encoding_for_model("gpt-4o")
tokens = enc.encode("helidannukelabekkallave")
print(tokens)
print([enc.decode([t]) for t in tokens])
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("albert-base-v2")

tokens = tokenizer.tokenize("helidannukelabekkallave")
ids    = tokenizer.encode("helidannukelabekkallave")

print(tokens)
print(ids)
print(tokenizer.decode(ids))

print(tokenizer.backend_tokenizer.model)
```

```python
# SentencePiece

from huggingface_hub import hf_hub_download
import sentencepiece as spm

# Download the model file from HuggingFace
model_path = hf_hub_download(
    repo_id="xlm-roberta-base",
    filename="sentencepiece.bpe.model"
)

sp = spm.SentencePieceProcessor()
sp.load(model_path)   # ← fixed: uses the downloaded file path

tokens = sp.encode("helidannukelabekkallave", out_type=str)
ids    = sp.encode("helidannukelabekkallave", out_type=int)

print(tokens)
print(ids)
print(sp.decode(ids))
```

```python
from transformers import AutoTokenizer

# BERT uses WordPiece
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

tokens = tokenizer.tokenize("helidannukelabekkallave")
ids    = tokenizer.encode("helidannukelabekkallave")

print(tokens)
print(ids)
print(tokenizer.decode(ids))

# Verify it's WordPiece
print(tokenizer.backend_tokenizer.model)
```
